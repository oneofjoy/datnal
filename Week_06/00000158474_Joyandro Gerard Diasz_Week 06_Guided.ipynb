### Scatter plots

Scatter plots show the relationship between two variables in the form of dots on the plot.

#### Scatter plots with Matplotlib

Matplotlib is at the base of most Python plotting packages.
<b>Note</b>: The IPyhon magic command %matplotlib inline enables the display of graphics inline with the Python code. If you do not include this command your graphs will not be displayed.

```python
import matplotlib.pyplot as plt
%matplotlib inline 
plt.plot(cars['Horsepower'], cars['Miles_per_Gallon'], 'ro')
```

### Scatter plots with Pandas. 

While you can create most any visualization using Matplotlib, with enough code, you may want a simpler approach if your data are in a Pandas data frame.

```python
cars.plot(kind = 'scatter', x = 'Horsepower', y = 'Miles_per_Gallon')
```

We can take advantage of the fact that Pandas plot methods are built on matplotlib. Thus, we can specify a figure and one or more axes within that figure. Then, we can control many plot attributes once we have specified the axes. 

```python
import matplotlib.pyplot as plt
fig = plt.figure(figsize=(6, 6)) # define plot area
ax = fig.gca() # define axis                   
cars.plot(kind = 'scatter', x = 'Horsepower', y = 'Miles_per_Gallon', ax = ax)
ax.set_title('Scatter plot of Horsepower vs Miles_per_Gallon') # Give the plot a main title
ax.set_xlabel('Horsepower') # Set text for the x axis
ax.set_ylabel('Miles_per_Gallon')# Set text for y axis
```

### Line plots

Line plots are similar to point plots. In line plots the discrete points are connected by lines.

```python
import pandas as pd
x = list(range(100))
y = [z * z for z in range(100)]
df = pd.DataFrame({'x':x, 'y':y})
```

```python
fig = plt.figure(figsize=(10, 10)) # define plot area
ax = fig.gca() # define axis                   
df.plot(x = 'x', y = 'y', ax = ax) ## line is the default plot type
ax.set_title('Line plot of x^2 vs. x') # Give the plot a main title
ax.set_xlabel('x') # Set text for the x axis
ax.set_ylabel('x^2')# Set text for y axis
```

### Case study

Membuat sebuah line chart dengan Matplotlib, yang menunjukkan jumlah pembeli harian (daily number of customers) selama bulan Desember.

```python
# Import library yang dibutuhkan
import datetime
data = pd.read_csv('retail.csv')
```

```python
# Buat kolom order_month
data['order_month'] = data['order_date'].apply(lambda x: datetime.datetime.strptime(x, "%Y-%m-%d").strftime('%Y-%m'))
```

```python
# Buat kolom gmv
data['gmv'] = data['item_price'] * data['quantity']
```

```python
# Plot grafik sesuai dengan instruksi
plt.figure(figsize=(10, 5))
data[data['order_month']=='2019-12'].groupby(['order_date'])['customer_id'].nunique().plot(color='red', marker='.', linewidth=2)
plt.title('Daily Number of Customers - December 2019', loc='center', pad=30, fontsize=20, color='violet')
plt.xlabel('Order Date', fontsize=15, color='blue')
plt.ylabel('Number of Customers', fontsize=15, color='blue')
plt.grid(color='darkgray', linestyle=':', linewidth=0.5)
plt.ylim(ymin=0)
plt.show()
```

```python
# Buat Multi-Line Chart
data.groupby(['order_month','brand'])['gmv'].sum().unstack().plot()
plt.title('Monthly GMV Year 2019 - Breakdown by Brand', loc='center', pad=30, fontsize=20, color='blue')
plt.xlabel('order Month', fontsize = 15)
plt.ylabel('Total Amount (in Billions)', fontsize = 15)
plt.grid(color='darkgray', linestyle=':', linewidth=0.5)
plt.ylim(ymin=0)
labels, locations = plt.yticks()
plt.yticks(labels, (labels/1000000000).astype(int))
plt.gcf().set_size_inches(10, 5)
plt.tight_layout()
plt.show()
```

### Customization

```python
import matplotlib.pyplot as plt
data.groupby(['order_month','brand'])['gmv'].sum().unstack().plot()
plt.title('Monthly GMV Year 2019 - Breakdown by Brand',loc='center',pad=30, fontsize=20, color='blue')
plt.xlabel('Order Month', fontsize = 15)
plt.ylabel('Total Amount (in Billions)',fontsize = 15)
plt.grid(color='darkgray', linestyle=':', linewidth=0.5)
plt.ylim(ymin=0)
labels, locations = plt.yticks()
plt.yticks(labels, (labels/1000000000).astype(int))
plt.legend(loc='right', bbox_to_anchor=(1.6, 0.5), shadow=True, ncol=2)
plt.gcf().set_size_inches(12, 5)
plt.tight_layout()
plt.show()
```

### Pie Chart

Membuat Subset Data

```python
dataset_dki = data[(data['province']=='DKI Jakarta') & (data['order_month'] >= '2019-10')]
print(dataset_dki.head())
```

```python
import matplotlib.pyplot as plt
gmv_per_city_dki= dataset_dki.groupby('city')['gmv'].sum().reset_index()
plt.figure(figsize=(6,6))
plt.pie(gmv_per_city_dki['gmv'], labels = gmv_per_city_dki['city'],autopct='%1.2f%%')
plt.title('GMV Contribution Per City - DKI Jakarta in Q4 2019',loc='center', pad=30, fontsize=15, color='blue')
plt.show()
```

### Bar plots

Bar plots are used to display the counts of unique values of a categorical variable. The height of the bar represents the count for each unique category of the variable.

```python
counts = cars['Origin'].value_counts() # find the counts for each unique category
counts
```

```python
fig = plt.figure(figsize=(6,6)) # define plot area
ax = fig.gca() # define axis    
counts.plot.bar(ax = ax) # Use the plot.bar method on the counts data frame
ax.set_title('Number of auto types by origin') # Give the plot a main title
ax.set_xlabel('Name') # Set text for the x axis
ax.set_ylabel('Number of autos')# Set text for y axis
```

Multi-bar chart 

```python
import matplotlib.pyplot as plt
dataset_dki.groupby(['city','order_month'])['gmv'].sum().unstack().plot(kind='bar')
plt.title('GMV Per City, Breakdown by Month\nDKI Jakarta in Q4 2019',loc='center',pad=30, fontsize=15, color='blue')
plt.xlabel('Province', fontsize = 12)
plt.ylabel('Total Amount (in Billions)', fontsize = 12)
plt.legend(bbox_to_anchor=(1, 1), shadow=True, title='Month')
plt.ylim(ymin=0)
labels, locations = plt.yticks()
plt.yticks(labels, (labels/1000000000).astype(int))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Stacked Chart

```python
import matplotlib.pyplot as plt
dataset_dki.groupby(['order_month','city'])['gmv'].sum().sort_values(ascending=False).unstack().plot(kind='bar', stacked=True)
plt.title('GMV Per Month, Breakdown by City\nDKI Jakarta in Q4 2019',
          loc='center',pad=30, fontsize=15, color='blue')
plt.xlabel('Order Month', fontsize = 12)
plt.ylabel('Total Amount (in Billions)', fontsize = 12)
plt.legend(bbox_to_anchor=(1, 1), shadow=True, ncol=1, title='City')
plt.ylim(ymin=0)
labels, locations = plt.yticks()
plt.yticks(labels, (labels/1000000000).astype(int))
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
     
```

### Histograms

Histograms are related to bar plots. Histograms are used for numeric variables. Whereas, a bar plot shows the counts of unique categories, a histogram shows the number of data with values within a bin.

```python
fig = plt.figure(figsize=(6,6)) # define plot area
ax = fig.gca() # define axis    
cars['Acceleration'].plot.hist(ax = ax) # Use the plot.hist method on subset of the data frame
ax.set_title('Histogram of auto acceleration') # Give the plot a main title
ax.set_xlabel('Acceleration') # Set text for the x axis
ax.set_ylabel('Number of autos')# Set text for y axis
```

### Box plots

Box plots are another way to visualize the distribution of data values.

```python
# Create a vertical box plot for 'quantity' and 'item_price'
plt.figure(figsize=(8, 6))
plt.boxplot([data['quantity'], data['item_price']], vert=True, labels=['Quantity', 'Item Price'])
plt.title('Vertical Box Plot of Quantity and Item Price')
plt.ylabel('Values')
plt.show()
```

### Kernel density plots

We will now start working with the Seaborn package. Seaborn is a newer Python package which abstracts lower level matplotlib charts. Saborn also implements some additional cutting-edge chart types.

Kernel density plots are similar in concept to a histogram. A kernel density plot displays the values of a smoothed density curve of the data values. In other words, the kernel density plot is a smoothed version of a histogram.

```python
import seaborn as sns
sns.set_style("whitegrid")
sns.kdeplot(cars['Acceleration'])
```

```python
fig = plt.figure(figsize=(6,6)) # define plot area
ax = fig.gca() # define axis 
sns.set_style("dark")
sns.kdeplot(cars['Acceleration'], ax = ax)
ax.set_title('KDE plot of auto Acceleration') # Give the plot a main title
ax.set_xlabel('Acceleration') # Set text for the x axis
ax.set_ylabel('Density')# Set text for y axis
```

```python
# Assuming 'cars' DataFrame is already loaded and contains 'Horsepower' and 'Name'
fig, ax = plt.subplots()  # define axis 
sns.set_style("whitegrid")

# Use a KDE plot for 'Horsepower' and ensure that 'Name' is not included since it's non-numeric
sns.kdeplot(data=cars['Horsepower'], ax=ax, color="blue", fill=True)
ax.set_title('KDE plot of Auto Horsepower')  # Give the plot a main title
ax.set_xlabel('Horsepower')  # Set text for the x axis

plt.show()
```

### Violin plots

A violin plot combines attributes of boxplots and a kernel density estimation plot. Like a box plot, the violin plots can be stacked, with a group by operation.

```python
# Create a violin plot to visualize the distribution of 'quantity' and 'item_price'
plt.figure(figsize=(10, 6))  # Define the plot size
sns.violinplot(data=data[['quantity', 'item_price']])  # Create violin plot
plt.title('Violin Plot of Quantity and Item Price')  # Set the title of the plot
plt.xlabel('Variable')  # Set the x-axis label
plt.ylabel('Value')  # Set the y-axis label
plt.show()  # Display the plot
```
