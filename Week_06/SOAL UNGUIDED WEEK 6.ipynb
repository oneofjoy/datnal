# Soal Unguided Data Analysis Week 6 Data Visualization

### 1. Import Dataset

Load Dataset sales_adidas_.csv, lihat informasi dan deskripsi dataset tersebut

### 2. Visualisasi Bar Plot

Visualisasikan total penjualan berdasarkan kategori produk dengan bar plot menggunakan library Seaborn (*tambahkan judul visualisasi menjadi 'Total Penjualan Berdasarkan Kategori Produk', kemudian nama sumbu x menjadi 'Produk' dan sumbu y menjadi 'Total Penjualan').

### 3. Visualisasi Pie Chart

Visualisasikan pie chart untuk melihat persentase Retailer yang ada dalam data dengan library Matplotlib (*tambahkan judul visualisasi menjadi 'Persentase Retailer'dan wajib menampilkan label persentase dengan dua digit desimal.

### 4. Visualisasi Line Chart

Visualisasikan Line Chart untuk melihat fluktuasi Total sales setiap periodenya.

### 5. Visualisasi Histogram

Visualisasikan histogram untuk melihat sebaran frekuensi dari Unit Sold (*tambahkan judul visualisasi menjadi 'Histogram Unit Terjual') dan tampilkan density curve nya.

### 6. Visualisasi Scatter Plot

Visualisasikan Scatter plot untuk melihat sebaran frekuensi dari Unit Sold (*tambahkan judul visualisasi menjadi 'Histogram Unit Terjual') dan tampilkan density curve nya.

### 7. Visualisasi Heat Map Correlation

Visualisasikan Heat Map untuk melihat nilai korelasi antar variabel (*pilih dan gunakan variabel numerik saja untuk menampilkan visualisasi).

### 8. Visualisasi Heat Map untuk Missing Values
