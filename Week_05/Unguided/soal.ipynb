## Soal 1: Membuat dan Menganalisis Korelasi, Distribusi Data, dan Box Plot

- ## a Membuat dan Menjelaskan Matriks Korelasi (Rubrik ke-1)
  **Deskripsi:**
Mulailah dengan memuat dataset winequality-red.csv. Lakukan analisis korelasi untuk semua variabel numerik dalam dataset ini. Cari tahu variabel mana yang memiliki korelasi terkuat dan terlemah terhadap kualitas anggur (`quality`). Untuk menghitung matriks korelasi, Anda bisa menggunakan metode `.corr()`. Agar lebih mudah melihat hubungan, urutkan hasil korelasi menggunakan `.sort_values(ascending=False)`. 

- ## b Membuat Visualisasi Distribusi Data dan Menjelaskan Skewness (Rubrik ke-2)
  **Deskripsi:**
Selanjutnya, pilihlah variabel `residual sugar` dan buatlah visualisasi distribusi datanya. Anda dapat menggunakan `sns.histplot()` dengan tambahan parameter `kde=True` untuk membuat histogram yang disertai garis KDE (Kernel Density Estimate). Setelah itu, hitunglah skewness dari distribusi `residual sugar` menggunakan `.skew()` dan jelaskan apakah distribusi tersebut condong ke kanan, kiri, atau normal. 

- ## c Membuat Box Plot Berdasarkan Kategori Kualitas (Rubrik ke-3)
  **Deskripsi:**
Kemudian, bagi data anggur ke dalam dua kategori berdasarkan nilai `quality`: anggap anggur dengan `quality` ≥ 6 sebagai "Baik" dan anggur dengan `quality` < 6 sebagai "Kurang". Buatlah visualisasi box plot untuk variabel `alcohol` dengan membandingkan kedua kategori tersebut. Analisis dan jelaskan apakah terdapat perbedaan signifikan dalam kandungan alkohol antara anggur berkualitas "Baik" dan "Kurang". Anda bisa menggunakan `sns.boxplot()` untuk membuat visualisasi ini. 

## Soal 2: Pembersihan Data, Visualisasi Distribusi, dan Analisis Outliers

- ## a Pembersihan Data (Duplikasi dan Missing Values) (Rubrik ke-4)
  **Deskripsi:**
Sekarang, muat dataset FuelConsumption.csv. Periksa apakah terdapat missing values atau duplikasi data dalam dataset tersebut. Jika ditemukan, bersihkan data dengan menghapus duplikasi dan mengisi missing values pada kolom numerik dengan nilai rata-rata kolom tersebut. Anda dapat memilih hanya kolom numerik menggunakan `select_dtypes(include='number')` sebelum mengisi missing values dengan metode `.mean()`. 

- ## b Visualisasi Distribusi Data Numerik (Rubrik ke-2)
  **Deskripsi:**
Setelah membersihkan data, buatlah ringkasan statistik deskriptif untuk setiap kolom numerik. Gunakan `.describe()` untuk mendapatkan ringkasan ini. Kemudian, visualisasikan distribusi dari setiap kolom numerik dengan membuat histogram yang disertai garis KDE. Gunakan `sns.histplot()` dan tampilkan hasilnya dalam format yang rapi. 

- ## c Deteksi Outliers dan Membuat Box Plot CO2 Emissions (Rubrik ke-4)
  **Deskripsi:**
Terakhir, deteksi outliers dalam dataset berdasarkan variabel `CO2EMISSIONS` dengan menggunakan metode Z-score. Setelah menghitung Z-score menggunakan `zscore()` dari `scipy.stats`, identifikasilah data yang memiliki Z-score di luar rentang -3 hingga 3 sebagai outliers. Visualisasikan hasil deteksi outliers ini dengan membuat box plot dari `CO2EMISSIONS` menggunakan `sns.boxplot()`. Diskusikan apakah outliers ini menunjukkan data yang ekstrem atau ada kemungkinan kesalahan data, serta bagaimana outliers ini dapat mempengaruhi analisis keseluruhan.
