```python
# Mengurutkan ListingDate dari paling awal
listing_ascending = daftar_saham.sort_values(
    by="ListingDate",
    ascending=True
)

print("Saham dengan Listing Date paling awal:")
display(
    listing_ascending[
        ["Code", "Name", "ListingDate"]
    ].head(1)
)


# Mengurutkan Code secara alfabet
code_ascending = daftar_saham.sort_values(
    by="Code",
    ascending=True
)

print("Code saham terakhir secara alfabet:")
display(
    code_ascending[
        ["Code", "Name"]
    ].tail(1)
)
```

```python
code_descending = daftar_saham.sort_values(
    by="Code",
    ascending=False
)

display(code_descending[["Code", "Name"]].head(1))
```
