```python
# Group by 'PRIMARY_TYPE' and calculate the total number of crimes and arrests
crime_grouped = data_kriminal.groupby('PRIMARY_TYPE').agg({
    'ID': 'count',            # Count the total number of crimes for each type
    'ARREST': 'sum'           # Sum the total number of arrests for each type
}).reset_index()

# Rename columns for clarity
crime_grouped.columns = ['Crime Type', 'Total Crimes', 'Total Arrests']

# Display the grouped data
crime_grouped.head()
```

```python
# Define specific location and year to filter (e.g., District 9 and Year 2016)
selected_district = 9
selected_year = 2016

# Filter the DataFrame based on the specified district and year
filtered_crimes = data_kriminal[(data_kriminal['DISTRICT'] == selected_district) & (data_kriminal['YEAR'] == selected_year)]

# Use groupby with agg on the filtered data
grouped_filtered_crimes = filtered_crimes.groupby('PRIMARY_TYPE').agg(
    Total_Crimes=('ID', 'count'),                # Count total number of crimes for each type
    Total_Arrests=('ARREST', 'sum'),             # Sum the total number of arrests for each type
    Average_Latitude=('LATITUDE', 'mean'),       # Calculate the average latitude for each crime type
    Average_Longitude=('LONGITUDE', 'mean')      # Calculate the average longitude for each crime type
).reset_index()

# Display the grouped data with aggregations
print(grouped_filtered_crimes)
```

```python
#merge based on assumed column names
kriminal = crime_grouped.merge(data_kriminal, 
                               left_on='Crime Type',  # Adjusted to correct column name in crime_grouped
                               right_on='PRIMARY_TYPE',  # Adjusted to correct column name in data_kriminal
                               suffixes=(False, False))
```

```python
# Find the total number of crimes recorded in the CRIME table.
cur.execute('SELECT COUNT(ID) FROM data_kriminal')
cur.fetchone()
```

```python
# List community areas with per capita income less than 11000.

cur.execute('''SELECT COMMUNITY_AREA_NAME FROM data_sensus
                WHERE PER_CAPITA_INCOME < 11000''')
#Convert result of Query to Pandas Dataframe and rename its Column
pd.DataFrame(cur.fetchall()).rename(columns={0:'COMMUNITY_AREA_NAME'})
```

```python
cur.execute(''' SELECT COMMUNITY_AREA_NAME, C FROM
                (SELECT C ,NUM FROM (SELECT COUNT(*) AS C,
                COMMUNITY_AREA_NUMBER AS NUM FROM data_kriminal
                GROUP BY COMMUNITY_AREA_NUMBER) ORDER BY C DESC LIMIT 2) A
                LEFT JOIN data_sensus B WHERE A.NUM = B.COMMUNITY_AREA_NUMBER ''')

pd.DataFrame(cur.fetchall()).rename(columns={0:'COMMUNITY_AREA_NAME',1:'Count Of Crimes'})
```
