```python
import pandas as pd
import numpy as np
from datetime import datetime

daftar_saham = pd.read_csv("DaftarSaham.csv")
financial = pd.read_csv("DataFinancialSaham.csv")
```

## <div align="center"> Question 01 </div>

```python
# Mengurutkan ListingDate dari paling awal
listing_ascending = daftar_saham.sort_values(
    by="ListingDate",
    ascending=True
)

print("Saham dengan Listing Date paling awal:")
display(
    listing_ascending[
        ["Code", "Name", "ListingDate"]
    ].head(1)
)


# Mengurutkan Code secara alfabet
code_ascending = daftar_saham.sort_values(
    by="Code",
    ascending=True
)

print("Code saham terakhir secara alfabet:")
display(
    code_ascending[
        ["Code", "Name"]
    ].tail(1)
)
```

```python
code_descending = daftar_saham.sort_values(
    by="Code",
    ascending=False
)

display(code_descending[["Code", "Name"]].head(1))
```

## <div align="center"> Question 02 </div>

```python
financial.columns = financial.columns.str.capitalize()

print(financial.columns)
```

```python
daftar_saham["ListingDate"] = pd.to_datetime(
    daftar_saham["ListingDate"],
    format="mixed"
)

print(daftar_saham["ListingDate"].dtype)
```

```python
daftar_filtered = daftar_saham[
    daftar_saham["ListingDate"]
    .dt.year
    .between(2020, 2023)
].copy()

display(
    daftar_filtered[
        ["Code", "Name", "ListingDate"]
    ].head()
)
```

```python
print("Jumlah saham:", len(daftar_filtered))

print(
    daftar_filtered["ListingDate"]
    .dt.year
    .value_counts()
    .sort_index()
)
```

## <div align="center"> Question 03 </div>

```python
sector_summary = (
    daftar_filtered
    .groupby("Sector")
    .agg(
        TotalShares=("Shares", "sum"),
        AverageMarketCap=("MarketCap", "mean")
    )
    .reset_index()
)

display(sector_summary)
```

## <div align="center"> Question 04 </div>

```python
tahun_sekarang = datetime.now().year

daftar_saham["TotalTahunOperasi"] = (
    tahun_sekarang
    - daftar_saham["ListingDate"].dt.year
)

display(
    daftar_saham[
        [
            "Code",
            "Name",
            "ListingDate",
            "TotalTahunOperasi"
        ]
    ].head()
)
```

## <div align="center"> Question 05 </div>

```python
merged_df = pd.merge(
    daftar_filtered,
    financial,
    left_on="Code",
    right_on="Symbol",
    how="inner"
)

display(merged_df.head())

print("Jumlah baris hasil merge:", len(merged_df))
print("Jumlah saham unik:", merged_df["Code"].nunique())
```

```python
merged_df["ListingYear"] = (
    merged_df["ListingDate"].dt.year
)
```

```python
merged_unique = merged_df.drop_duplicates(
    subset=["Code"]
)
```

```python
average_last_price = (
    merged_unique
    .groupby(
        ["Sector", "ListingYear"]
    )["LastPrice"]
    .mean()
    .reset_index(
        name="AverageLastPrice"
    )
)

average_last_price["AverageLastPrice"] = (
    average_last_price["AverageLastPrice"]
    .round(2)
)

display(average_last_price)
```
