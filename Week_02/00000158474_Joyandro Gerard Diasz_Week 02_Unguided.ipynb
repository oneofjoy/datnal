##### Problem 01

```python
data_mahasiswa = {
    "Mahasiswa A": {"jurusan": "Teknik Informatika", "SKS": 120},
    "Mahasiswa B": {"jurusan": "Teknik Informatika", "SKS": 90},
    "Mahasiswa C": {"jurusan": "Teknik Informatika", "SKS": 75}
}


def hitung_total_SKS(data_mahasiswa):
    total_sks = 0

    for nama, data in data_mahasiswa.items():
        total_sks += data["SKS"]

    return total_sks


def kategori_SKS(data_mahasiswa):
    hasil = {}

    for nama, data in data_mahasiswa.items():
        sks = data["SKS"]

        if sks >= 120:
            kategori = "Pass"
        elif sks >= 80:
            kategori = "Towards Pass"
        else:
            kategori = "Initial Stage"

        hasil[nama] = kategori

    return hasil
```

```python
print("Total SKS:", hitung_total_SKS(data_mahasiswa))
print("Kategori:", kategori_SKS(data_mahasiswa))
```

```python
def hitung_nilai_akhir(nilai_tugas, nilai_UTS, nilai_UAS):
    nilai_akhir = (
        nilai_tugas * 0.30 +
        nilai_UTS * 0.30 +
        nilai_UAS * 0.40
    )

    return nilai_akhir


nilai_akhir = hitung_nilai_akhir(80, 85, 90)
```

##### Problem 04

```python
while True:

    nama_baru = input(
        "Masukkan nama mahasiswa (atau ketik 'berhenti' untuk keluar): "
    )

    if nama_baru.lower() == "berhenti":
        break

    # Validasi nama
    if not nama_baru.replace(" ", "").isalpha():
        print("Nama tidak valid. Masukkan nama yang benar.")
        continue

    jurusan_baru = input("Masukkan jurusan: ")

    # Validasi SKS
    try:
        sks_baru = int(input("Masukkan jumlah SKS: "))

        if sks_baru <= 0:
            print("SKS harus positif.")
            continue

    except ValueError:
        print("Masukkan jumlah SKS yang valid.")
        continue

    data_mahasiswa[nama_baru] = {
        "jurusan": jurusan_baru,
        "SKS": sks_baru
    }

    print("Data mahasiswa berhasil ditambahkan.")


print("\nData mahasiswa:")
print(data_mahasiswa)
```

##### Problem 05

```python
def rata_rata_SKS_per_jurusan(data_mahasiswa):

    total_sks = {}
    jumlah_mahasiswa = {}

    # Mengelompokkan total SKS dan jumlah mahasiswa
    for nama, data in data_mahasiswa.items():

        jurusan = data["jurusan"]
        sks = data["SKS"]

        if jurusan not in total_sks:
            total_sks[jurusan] = 0
            jumlah_mahasiswa[jurusan] = 0

        total_sks[jurusan] += sks
        jumlah_mahasiswa[jurusan] += 1

    # Menghitung rata-rata
    rata_rata = {}

    for jurusan in total_sks:
        rata_rata[jurusan] = (
            total_sks[jurusan] /
            jumlah_mahasiswa[jurusan]
        )

    return rata_rata


hasil = rata_rata_SKS_per_jurusan(data_mahasiswa)

for jurusan, rata in hasil.items():
    print(jurusan, ":", rata)
```

##### Problem 06

```python
def analisis_mahasiswa(data_mahasiswa):

    def hitung_total_SKS():
        total = 0

        for nama, data in data_mahasiswa.items():
            total += data["SKS"]

        return total


    def rata_rata_SKS():
        if len(data_mahasiswa) == 0:
            return 0

        total = hitung_total_SKS()

        return total / len(data_mahasiswa)


    def cari_mahasiswa_terbaik():
        mahasiswa_terbaik = None
        sks_tertinggi = -1

        for nama, data in data_mahasiswa.items():

            if data["SKS"] > sks_tertinggi:
                sks_tertinggi = data["SKS"]
                mahasiswa_terbaik = nama

        return mahasiswa_terbaik, sks_tertinggi


    total = hitung_total_SKS()
    rata_rata = rata_rata_SKS()
    terbaik, sks = cari_mahasiswa_terbaik()

    print("Total SKS:", total)
    print("Rata-rata SKS:", rata_rata)
    print("Mahasiswa terbaik:", terbaik)
    print("SKS tertinggi:", sks)


analisis_mahasiswa(data_mahasiswa)
```

##### Problem 07

```python
def analisis_kehadiran(kehadiran):

    total_kehadiran = 0

    for nilai in kehadiran:
        total_kehadiran += nilai

    total_hari = len(kehadiran)

    if total_hari == 0:
        return "Data kehadiran kosong"

    persentase_kehadiran = total_kehadiran / total_hari

    print("Persentase Kehadiran:", persentase_kehadiran, "%")

    if persentase_kehadiran >= 75:
        return "Lulus"
    else:
        return "Tidak Lulus"


status = analisis_kehadiran([80, 85, 70, 90])

print(status)
```
